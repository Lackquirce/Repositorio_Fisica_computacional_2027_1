### **1. Coordenadas polares y cartesianas:** 
#### Supongamos que se nos da (es decir, el usuario introduce) la posición de un punto en un espacio bidimensional:

#### **a)** Si el punto se especifica  en coordenadas polares $r$,$\theta$. Escribe un programa (en `python` sin usar la biblioteca `numpy`) que calcule e imprima la conversión a coordenadas cartesianas $x$, $y$.

$\text{La regla de transformación de coordenadas polares a cartesianas está dada por: } $

$x = r\cos(\theta)$

$y = r\sin(\theta)$

```python
import math as mt

r = float(input("¿Cuál es tu coordenada radial?"))
θ = float(input("Ahora, ¿qué ángulo tiene dicha coordenada?" ))

while r<0:
    print("r debe ser una cantidad positiva")
    r = float(input("¿Cuál es tu coordenada radial?"))
    θ = float(input("Ahora, ¿qué ángulo tiene dicha coordenada?" ))

x = r*mt.cos(θ)
y = r*mt.sin(θ)

print ("Tu punto en coordenadas cartesianas es (x,y) = (",round(x,3),",",round(y,3),")")
```

#### **b)** Si el punto se especifica en coordenadas cartesianas $x$, $y$. Escribe un programa (en `python` sin usar `numpy`) que calcule e imprima la conversión a coordenadas polares $r$,$\theta$.

$\text{Por otra parte la trasnformación de coordenadas de cartesianas a polares está dada por: }$

$r = \sqrt{x^2 + y^2}$

$θ = \arctan{\left(\frac{y}{x}\right)}$

```python
import math as mt

x_0 = float(input("Este es el valor de la abscisa: "))
y_0 = float(input("Y el valor de la ordenada es: "))

r_0 = mt.sqrt(x_0**2+y_0**2)
θ_0 = mt.atan2(y_0,x_0) #Aunque no esté definido en el origen de manera formal, mt.atan2 lo toma en cuenta y devuelve un ángulo.

print("Tu punto en coordenadas polares es (r,θ) = (",r_0,",", round(θ_0,3),")")
```

### **2. Caída de una pelota desde una torre:** 
#### Se deja caer una pelota desde una torre de altura $h$. Tiene velocidad inicial cero y acelera hacia abajo con la gravedad.

Escribe un programa que le pida al usuario que ingrese la altura en metros de la torre y luego calcule e imprima el tiempo en segundos hasta que la pelota toque el suelo (ignora la resistencia del aire).

Después, usa tu programa para calcular el tiempo de una pelota lanzada desde una torre de $100m$ de altura.\\


$\text{Un cuerpo en caída libre sobre la Tierra y con resistencia del aire despreciable está descrito por la siguiente ecuación:}$

$h = v_0t + \frac{1}{2}gt^2$

$\text{donde\,\,\,\,} h \text{\,\,\,\,es la altura desde la cual se deja caer,\,\,\,\,} v_0 \text{\,\,\,\,su velocidad inicial y\,\,\,\,} t \text{\,\,\,\,el tiempo que tarda en llegar al fondo.}$
$\text{cuando\,\,\,\,} v_0 = 0 \text{\,\,\,\, despejando el tiempo de llegada al suelo se obtiene que:}$
$t = \sqrt{\frac{2h}{g}}$

```python
import numpy as np

g = 9.81 #Se va a considerar que está en [m/s^2]
h = float(input("Detalle desde qué altura (en metros) desea dejar caer la pelota :"))

while h<0:
    print("No hay alturas negativas a las que puedas dejar caer la pelota, considerando que el suelo está en el cero del sistema de referencia.")
    h = float(input("Detalle nuevamente a qué altura (en metros) desea dejar caer la pelota"))

if h == 0:
    print("La pelota ya está en el suelo")
else:
    t = np.sqrt(2*h/g)
    print("La pelota tardó en caer", round(t,3), "[s]")
```

### **3. Relatividad especial:** 
#### Una nave espacial viaja desde la Tierra en línea recta a una velocidad relativista $v$ a otro planeta a $x$ años luz de distancia. 

#### Escribe un programa que le pida al usuario el valor de $x$ y la velocidad $v$ como una fracción de la velocidad de la luz $c$, y que imprima el tiempo en años que tarda la nave espacial en llegar a su destino

#### **(a)** en el marco de reposo de un observador en la Tierra y


$\text{La velocidad de la luz por definición es\,\,\,} c = 299,792,458 [\frac{m}{s}]$ 
$\text{Con apoyo de la varianza del intervalo se puede mostrar que las nociones de tiempo entre dos sistemas de referencia está dado por:}$

$\Delta t = \frac{1}{\sqrt{1-\frac{v^2}{c^2}}}\Delta t'$

```python
import numpy as np

x_0 = np.abs(float(input("¿A cuántos años luz de distancia se encuentra el planeta que se va a visitar?")))
v_0 = np.abs(float(input("¿Qué fracción de la velocidad de la luz lleva la nave espacial? Recuerda que es un valor entre 0 y 1:")))

while v_0>1:
    print("La nave no puede superar la velocidad de la luz.")
    v_0 = np.abs(float(input("¿Qué fracción de la velocidad de la luz lleva la nave espacial?")))

if v_0==0:
    print("La nave está en reposo, jamás vas a llegar al planeta.")
else:
    t_0 = x_0/v_0
    print("La nave tardará ", round(t_0,3), "[años] en llegar al planeta, visto desde la Tierra como sistema de referencia inercial en reposo.")
```

### **4. Paso de potencial cuántico.** 
#### Un conocido problema de mecánica cuántica involucra una partícula de masa $m$ que encuentra un paso de potencial unidimensional, como este:

<center><img src="img/qstep1.png" width="40%" class="center"/></center>

#### La partícula con energía cinética inicial $E$ y vector de onda $k_1=\sqrt{2mE}/\hbar$ entra por la izquierda y encuentra un salto repentino en la energía potencial de altura $V$ en la posición $x=0$. Resolviendo la ecuación de Schrödinger, se puede demostrar que cuando $E>V$ la partícula puede 

#### (a) **pasar el escalón**, en cuyo caso tiene una energía cinética menor a $E-V$ en el otro lado y un vector de onda correspondientemente menor a $k_2=\sqrt{2m(E-V)}/\hbar$, \; o bien (b) **reflejarse**, manteniendo toda su energía cinética y un vector de onda sin cambios, pero moviéndose en la dirección opuesta. Las probabilidades $T$ y $R$ de transmisión y reflexión están dadas por:

<h3 align="center">$T = {4k_1k_2\over(k_1+k_2)^2}\,,\qquad$
$R = \biggl( {k_1-k_2\over k_1+k_2} \biggr)^2.$</h3>

#### Supongamos que tenemos una partícula con una masa igual a la masa del electrón. $m=9.11\times10^{-31}\,$kg y energía de $10\,$eV, al encontrar un escalón de potencial de altura $9\,$eV. Escribe un programa en Python para calcular e imprimir las probabilidades de transmisión y reflexión utilizando las fórmulas anteriores.


```python
import numpy as np

m = 9.11e-31
E = 10
V = 9
ħ = 6.582e-16 #En eV*s

k_1 = np.sqrt(2*m*E)/ħ
k_2 = np.sqrt(2*m*(E-V))/ħ

T = np.divide(4*k_1*k_2,(k_1+k_2)**2)
R = (np.divide(k_1-k_2,k_1+k_2))**2

print("La probabilidad de que la partícula se transmita es de T =",round(T,3), ", mientras que la probalidad de que se refleje es de R =", round(R,3))
print("Esto además cumple que T + R = ", round(T+R,3))
```
