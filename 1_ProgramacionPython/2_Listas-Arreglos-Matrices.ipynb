#### La función `map`


Una función especialmente útil para las listas es `map`, que permite aplicar funciones ordinarias, como `log`  o `sqrt`, a todos los elementos de una lista a la vez. Así, `map(log,r)` toma el logaritmo natural de cada elemento de la lista `r`:

```python
r = [ 1.0, 1.5, 2.2, mt.e ]
map(mt.log,r) #Crea un objeto, pero hay que transformarlo en lista.
```

```python
logr = list(map(mt.log,r)) 
```

```python
print(logr) #Con lo anterior define a una lista a partir del objeto creado con map al cual se le aplicó el logaritmo a cada elemento.
```

```python
for i in map(mt.log,r):
    print(i)
```

```python
p = [1.1, 1.5, 3.5, 4.99]
p_int = list(map(round,p))
p_factorial = list(map(mt.factorial,p_int))
print("Lista redondeada en enteros",p_int,"y el factorial del redondeado",p_factorial)
```
