EJERCICIO 3. Coeficientes binomiales.

El coeficiene binomial 
$ \binom{n}{k} = \frac{n!}{k!(n-k)!}
$


```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n*factorial(n-1)
```

```python
def binomial(n,k):
    if k == 0 or n == k:
        return 1
    if k>= 1 and n > k:
        return int(factorial(n)/(factorial(k)*factorial(n-k)))
    else:
        print("k no puede ser menor que n")
```

```python
binomial(4,10)
```
