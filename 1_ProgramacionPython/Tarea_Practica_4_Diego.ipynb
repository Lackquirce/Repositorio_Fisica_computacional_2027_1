````python
t,v = [],[]
with open("data/velocidades.txt", "r") as archivo:
    for linea in archivo:
        separacion = linea.split()
        t.append(int(separacion[0]))
        v.append(float(separacion[1]))
````

````python
def velocidad(t):
    if t < 0 or t > 100:
        raise ValueError("Solo está definida para los primeros 100 segundos.")
    else:
        return v[int(t)]

def Trapecio(f,a,b,N):
    h = (b-a)/N

    Integral = 0
    Int1 = (f(a)+f(b))/2

    for i in range(1, N):
        Integral += f(a+i*h)

    I = h*(Integral + Int1)
    return I

x = [0.0]
for j in range(1,len(t)):
    x_j  = Trapecio(velocidad,0,j,j)
    x.append(round(x_j,3))
print("La siguiente tabla muestra la posición de la partícula como función del tiempo.")
print("Tiempo [s] --- Posición [m]")
for i in range(len(t)):
    print(t[i], "     ---     ", x[i])
````

````python
import numpy as np

def simpson1(f,a,b,n):
    if n%2 != 0 or n < 0:
        raise ValueError("Este método requiere de una cantidad positiva par de puntos en la refinación del intervalo.")
    h = (b-a)/n
    suma1 = 0
    suma2 = 0
    N = int(n/2)
    for k in range(1,N+1):
        suma1 += f(a+(2*k-1)*h)
    for j in range(1,N):
        suma2 += f(a+(2*j*h))
    IntegralAprox = (1/3)*h*((f(a) + f(b)) + 4*suma1 + 2*suma2)
    return IntegralAprox
    


def BesselJ(m,x):
    if m < 0:
        raise ValueError("Las funciones de Bessel de orden m de primer tipo solo admiten enteros no negativos.")
    elif isinstance(m,float):
        raise TypeError("No hay ordenes decimales para las funciones de Bessel de primer tipo.")
    elif x < 0:
        raise ValueError("Solo están definidos para argumentos no negativos.")
    def integrando(θ):
        return np.cos(m*θ - x*np.sin(θ))
    return (1/np.pi)*simpson1(integrando,0,np.pi,1000)
    
````

### 3. Capacidad calorífica de un sólido

La teoría de sólidos de Debye dice que la capacidad calorífica de un sólido a una temperatura $T$ está dada por:

$$C_V = 9V \rho k_B \left(\frac{T}{\theta_D}\right)^3 \int_0^{\theta_D/T} \frac{x^4 e^x}{(e^x - 1)^2} dx $$

donde $V$ es el volumen del sólido, $\rho$ es la densidad numérica de los átomos, $k_B$ es la constante de Boltzmann y $\theta_D$ es la llamada temperatura de Debye, una propiedad de sólidos que depende de su densidad y la velocidad del sonido.

**(a)** Escribe una función $C_V(T)$ que calcule la $C_V$ para un valor dado de temperatura, de una muestra que consta de 1000 centímetros cúbicos de aluminio sólido, con densidad numérica de $\rho = 6.022 	\times 10^{28}\text{ m}^{-3}$ y una temperatura de Debye de $\theta_D = 428	\text{ K}$. Utiliza la cuadratura gaussiana para evaluar la integral, con $N=50$ puntos muestra.

````python
import scipy as sp
import numpy as np

def gauss_xi_w_i(n,a,b):

    if a > b:
        raise ValueError("Esta funnción requiere que sea a menor que b ya que sigue la estructura de un intervalo cerrado [a,b].")
    
    raices = sp.special.roots_legendre(n)[0]
    pesos = sp.special.roots_legendre(n)[1]
    xi = (b-a)/2 * raices + 1/2 * (b+a)
    wi = (b-a)/2 * pesos
    return xi, wi
    

def CGauss1D(n,f,a,b):
    C = 0
    xk = gauss_xi_w_i(n,a,b)[0]
    wk = gauss_xi_w_i(n,a,b)[1]

    for j in range(0,n):
        C += wk[j]*f(xk[j])
    return float(C)
````

### 4. Incertidumbre cuántica en el oscilador armónico

En unidades donde todas las constantes son 1, la función de onda del n-ésimo nivel de energía del oscilador armónico cuántico unidimensional, (i.e., una partícula puntual sin espín en un pozo de potencial cuadrático) está dada por:

$$ \psi_n(x) = \frac{1}{\sqrt{2^n n! \sqrt{\pi}}} e^{-x^2/2} H_n(x) $$

para $n=0...\infty$, donde $H_n(x)$ es el n-ésimo polinomio de Hermite. Los polinomios de Hermite satisfacen una relación algo similar a la de los números de Fibonacci, aunque más compleja:

$$ H_{n+1}(x) = 2x H_n(x) - 2n H_{n-1}(x) $$

Los dos primeros polinomios de Hermite son $H_0(x)=1$ y $H_1(x)=2x$.

**(a)** Escribe una función $H(n,x)$ que calcule $H_n(x)$ dados $x$ y cualquier número entero $n \ge 0$. Utiliza tu función para hacer una gráfica que muestre las funciones de onda del oscilador armónico para $n=0, 1, 2$ y $3$, todos en el mismo gráfico, en el rango de $x=-4$ a $x=4$. *Hint:* Hay una función factorial en el paquete `math` que calcula el factorial de un número entero.

````python
def Hermite(n,x):
    if not isinstance(n, int):
        raise TypeError("Esta función solo está definida para n enteros no negativos.")
    if n < 0:
        raise ValueError("Esta función de Hermite solo está definida para n enteros no negativos.")
    if n == 0:
        return 1
    if n == 1:
        return 2*x
    hn_2 = 1
    hn_1 = 2*x
    for m in range(2, n + 1):
        hn= 2*x*hn_1-2*(m-1)*hn_2
        hn_2 = hn_1
        hn_1 = hn
    return hn
````

### 5. Atracción gravitacional de una lámina uniforme

Una lámina de metal cuadrada uniforme flota inmóvil en el espacio:


La lámina mide 10 m de lado, tiene un espesor insignificante y una masa de 10 toneladas.

**(a)** Considera la fuerza gravitacional debida a la placa que siente una masa puntual de 1 kg a una distancia $z$ del centro del cuadrado, en dirección perpendicular a la lámina, como se muestra en la figura. Demuestra que la componente de la fuerza a lo largo del eje z es:

$$ F_z = G\sigma z \iint_{-L/2}^{L/2} \frac{dx dy}{(x^2 + y^2 + z^2)^{3/2}} $$

donde $G = 6.674 	\times 10^{-11}	\text{ m}^3	\text{ kg}^{-1}	\text{ s}^{-2}$ es la constante gravitacional de Newton y $\sigma$ es la masa por unidad de área de la hoja.





Sea una lámina cuadrada uniforme de lado $L$, centrada en el origen y contenida en el plano $xy$. Su densidad superficial es

$$\sigma=\frac{M}{L^2}.$$

Para un elemento diferencial de área $dA=dx\,dy$, la masa es

$$\boxed{dm=\sigma\,dx\,dy}.$$


La masa de prueba $m=1\,\mathrm{kg}$ se encuentra en $(0,0,z)$, mientras que $dm$ se localiza en $(x,y,0)$.

El vector que apunta desde la masa de prueba hacia el elemento diferencial es

$$\mathbf r=-x\hat{\mathbf i}-y\hat{\mathbf j}-z\hat{\mathbf k},$$

con magnitud

$$r=\sqrt{x^2+y^2+z^2}.$$

Por tanto,

$$\boxed{r^2=x^2+y^2+z^2}.$$


La fuerza diferencial ejercida sobre la masa de prueba es

$$d\mathbf F=-G\frac{m\,dm}{r^3}\mathbf r.$$

Sustituyendo $m=1\,\mathrm{kg}$ y $dm=\sigma\,dx\,dy$:

$$d\mathbf F=-G\sigma\frac{-x\hat{\mathbf i}-y\hat{\mathbf j}-z\hat{\mathbf k}}{(x^2+y^2+z^2)^{3/2}}\,dx\,dy.$$

Así, la componente vertical resulta

$$\boxed{dF_z=-G\sigma z\frac{dx\,dy}{(x^2+y^2+z^2)^{3/2}}}.$$

Por simetría de la lámina,

$$\iint_S dF_x=0,\qquad\iint_S dF_y=0.$$

Por lo tanto,

$$\mathbf F=F_z(-\hat{\mathbf k}),$$
para $z>0$.

Integrando la componente vertical sobre toda la lámina,

$$F_z=\iint_S dF_z.$$

Como $G$, $\sigma$ y $z$ son constantes respecto a $x$ e $y$,

$$F_z=G\sigma z\int_{-L/2}^{L/2}\int_{-L/2}^{L/2}\frac{dx\,dy}{(x^2+y^2+z^2)^{3/2}}.$$

Por tanto, la expresión para la componente vertical de la fuerza gravitacional es

$$\boxed{F_z=G\sigma z\int_{L/2}^{L/2}\int_{-L/2}^{L/2}\frac{dx\,dy}{(x^2+y^2+z^2)^{3/2}}}.$$

La magnitud de la fuerza es $F=|F_z|$, y su dirección es perpendicular a la lámina, hacia ella.

**(b)** Escribe un programa para calcular y graficar la fuerza en función de $z$ de $z=0$ hasta $z=10	\text{ m}$. Para la integral doble utiliza la cuadratura Gaussiana (doble):

$$ I \simeq \sum_{i=1}^N \sum_{j=1}^N \omega_i \omega_j f(x_i, x_j) $$

con 100 puntos de muestra a lo largo de cada eje.

````python
import scipy as sp
import numpy as np

def gauss_xi_w_i(n,a,b):

    if a > b:
        raise ValueError("Esta funnción requiere que sea a menor que b ya que sigue la estructura de un intervalo cerrado [a,b].")
    
    raices = sp.special.roots_legendre(n)[0]
    pesos = sp.special.roots_legendre(n)[1]
    xi = (b-a)/2 * raices + 1/2 * (b+a)
    wi = (b-a)/2 * pesos
    return xi, wi
    

def CGauss2D(n,f,a,b,c,d):
    C = 0
    xi,wi = gauss_xi_w_i(n,a,b)
    yj,wj = gauss_xi_w_i(n,c,d)

    
    for i in range(0,n):
        for j in range(0,n):
            C += wi[i]*wj[j]*f(xi[i],yj[j])
    return float(C)
````

*Este cálculo puede considerarse como un modelo de la atracción gravitacional de una galaxia. La mayor parte de la masa de una galaxia espiral (como nuestra Vía Láctea) se encuentra en un plano o disco delgado que pasa por el centro galáctico, y la atracción gravitacional ejercida por ese plano sobre cuerpos fuera de la galaxia se puede calcular con los métodos que hemos empleado aquí.*

El efecto aparece porque el método de la cuadratura de Gauss con 100 puntos crea una cuadrícula de evaluación discreta. Al calcular una integral falsamente pequeña y multiplicarla por un factor $z$ que tiende a cero, el resultado de la fuerza colapsa artificialmente.

Una alternativa para disminuir el artefacto numérico es utilizar coordenadas polares, ya que el integrando depende de $x^2+y^2$. ### 6. Rejillas de difracción

La luz con longitud de onda $\lambda$ incide sobre una rejilla de difracción de ancho total $w$, se difracta, se enfoca con una lente convergente de distancia focal $f$ y proyecta sobre una pantalla:


La teoría nos dice que la intensidad del patrón de difracción en la pantalla, a una distancia $x$ del eje central del sistema, está dada por:

$$ I(x) = \left| \int_{-w/2}^{w/2} \sqrt{q(u)} e^{i\frac{2\pi x u}{\lambda f}} du 
\right|^2 $$

donde $q(u)$ es la función de transmisión de intensidad de la rejilla de difracción a una distancia $u$ del eje central, i.e., la fracción de la luz incidente que deja pasar la rejilla.

**(a)** Considera una rejilla con función de transmisión $q(u) = \sin^2(\alpha u)$. ¿Cuál es la separación de las "rendijas" en esta rejilla, expresada en términos de $\alpha$?


---
Para determinar la separación entre las rendijas, debemos encontrar el período espacial de la función de transmisión de intensidad:

$$q(u)=\sin^2(\alpha u).$$

Utilizando la identidad trigonométrica

$$\sin^2(\alpha u)=\frac{1-\cos(2\alpha u)}{2},$$

observando que el período de la función está determinado por el argumento del coseno. Como el período del coseno es $2\pi$, la separación $d$ entre dos máximos consecutivos satisface que:

$$2\alpha(u+d)-2\alpha u=2\pi$$

Simplificando la expresión anterior, obtenemos

$$2\alpha d=2\pi$$

Por lo tanto, despejando $d$, resulta

$$\boxed{d=\frac{\pi}{\alpha}}$$ suponiendo que $\alpha>0$.

````python
import numpy as np

def q(u):
    d = 20e-6
    alpha = np.pi/d
    return (np.sin(alpha*u))**2
````
