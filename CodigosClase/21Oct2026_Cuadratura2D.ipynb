```python
def f6(x,mu=0,s=1):
    return (1/(np.sqrt(2*np.pi*s**2)))*np.e**(-(1/2)*((x-mu)/s)**2)

def f(x,y):
    return x**2+y**2 + 7 
```

```python
a = -10
b = 10
N = 100
c = -2
d = 2
```

```python
def gauss_xw_ab(N,a,b):
    x,w = sp.special.roots_legendre(N)
    return 0.5*(b-a)*x+0.5*(b+a),0.5*(b-a)*w

def CuadGauss(N,f,a,b):
    S = 0
    
    x,w = gauss_xw_ab(N,a,b)
    
    for i in range(N):
        S = S + w[i]*f(x[i])
        
    return float(S)


def CuadGauss2D(N,f,a,b,c,d):
    S = 0
    
    x,wx = gauss_xw_ab(N,a,b)
    y,wy = gauss_xw_ab(N,c,d)
    
    for i in range(N):
        for j in range(N):
            S = S + wx[i]*wy[j]*f(x[i],y[j])
        
    return float(S)
```

```python
I = CuadGauss2D(N,f,a,b,c,d)
```
